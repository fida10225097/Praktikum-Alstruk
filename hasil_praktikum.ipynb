```python
# Mengenali Tipe Data
nama = "Fajar"
umur = 19
berat = 54.5
terdaftar = True

print(nama)
print(umur)
print(berat)
print(terdaftar)

print(type(nama))
print(type(umur))
print(type(berat))
print(type(terdaftar))

# mengubah umur menjadi 20
umur = 20
print("umur setelah diubah: ", umur)

# Operator Aritmatika
harga = 12500
jumlah = 4

# menghitung total harga
total_harga = harga * jumlah

# menghitung sisa pembagian dengan operator %
sisa = total_harga % 2000

# menghitung harga setelah diskon
diskon = 5000
harga_setelah_diskon = total_harga - diskon

# menampilkan hasil perhitungan
print("\nHarga buku per buah: Rp", harga)
print("Jumlah buku:", jumlah)
print("Total Harga: Rp", total_harga)
print("Sisa pembagia total harga 2000:", sisa)
print("Diskon: Rp", diskon)
print("Harga setelah diskon: Rp", harga_setelah_diskon)

# Operator Perbandingan dan Logika
nilai = 82
kehadiran = 90

# memeriksa masing-masing syarat
syarat_nilai = nilai >= 75
syarat_kehadiran = kehadiran >= 80

print("\nNilai mahasiswa:", nilai)
print("Kehadiran mahasiswa:", kehadiran)
print("Memenuhi syarat nilai:", syarat_nilai)
print("Memenuhi syarat kehadiran:", syarat_kehadiran)

# menggunakan operator and
print("Lulus menggunakan and:", syarat_nilai and syarat_kehadiran)

# mengubah nilai menjadi 70
nilai = 70
syarat_nilai = nilai >= 75

print("Nilai setelah diubah:", nilai)
print("Memenuhi syarat nilai:", syarat_nilai)
print("Lulus menggunakan and:", syarat_nilai and syarat_kehadiran)

# menggunakan operator or
print("Lulus menggunakan or:", syarat_nilai or syarat_kehadiran)

# Operator Penugasan
poin = 100
print("\nNilai awal:", poin)

# menambahkan 25
poin += 25
print("Setelah ditambah 25:", poin)

# mengurangi 10 poim
poin -= 10
print("Setelah dikurangi 10:", poin)

# mengalikan poin dengan 2
poin *= 2
print("Setelah dikali 2:", poin)

# membagi poin dengan 2
poin /= 2
print("Setelah dibagi 2:", poin)

# Sturktur Data Dasar
nilai = [78, 85, 90, 85]
mahasiswa = {"nama": "Nadia", "kelas": "IF-A"}
koordinat = (5, 8)
angka = {2, 4, 4, 6, 8}

# menampilkan nilai pertama dari list
print("\nNilai pertama:", nilai[0])

# menambahkan nilai 95 ke list
nilai.append(95)
print("List terbaru:", nilai)

# menampilkan nama mahasiswa dictionary
print("Nama mahasiswa:", mahasiswa["nama"])

# menampilkan koordinat pertama dari tuple
print("Koordinat pertama:", koordinat[0])

# menampilkan angka unik yang sudah diurutkan
print("Angka unik:", sorted(angka))

# menampilkan jumlah angka 4 dalam set
print("Angka 4 dalam set:", angka)

# Tantangan Terintegrasi - Program Rekap Nilai
nama = "Fida Maulida"
nilai_tugas = 85
nilai_ujian = 87

# menghitung rata-rata
rata_rata = (nilai_tugas + nilai_ujian) / 2

# status kelulusan
lulus = rata_rata >= 75

# menampilkan hasil
print("\nNama mahasiswa:", nama)
print("Nilai tugas:", nilai_tugas)
print("Nilai ujian:", nilai_ujian)
print("Rata-rata nilai:", rata_rata)
print("Status lulus:", lulus)
```
